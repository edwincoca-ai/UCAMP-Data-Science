# Reto 1 — Creación de una cuenta bancaria

## Descripción

Se desarrollará una clase llamada `Cuenta`.

Cada vez que un cliente abra una cuenta nueva,
el saldo inicial será de $0.

La clase contará con dos métodos principales:

- `ingreso()`: permitirá depositar dinero.
- `egreso()`: permitirá retirar dinero.

El sistema verificará que los retiros no superen
el saldo disponible.

Si el cliente intenta retirar más dinero del que
tiene, se mostrará una alerta.


```python
# Crear la clase Cuenta
class Cuenta:

    # Constructor de la cuenta
    def __init__(self):
        self.saldo = 0
        print(f'Tu saldo es: {self.saldo}')

    # Método para depositar dinero
    def ingreso(self, ingreso):
        self.saldo = self.saldo + ingreso
        print(f'Tu saldo es: {self.saldo}')

    # Método para retirar dinero
    def egreso(self, egreso):
        if egreso > self.saldo:
            print("No se tiene saldo suficiente")
        else:
            self.saldo = self.saldo - egreso
            print(f'Tu saldo es: {self.saldo}')
```

## Pruebas del sistema bancario

Se realizarán cuatro pruebas para verificar
el funcionamiento de la clase Cuenta.

1. Crear una cuenta con saldo inicial de $0.
2. Depositar $100.
3. Intentar retirar $150.
4. Retirar $50.

El sistema deberá impedir el retiro de $150
porque supera el saldo disponible.

```python
# Crear una cuenta nueva
Alex = Cuenta()

# Depositar $100
Alex.ingreso(100)

# Intentar retirar $150
Alex.egreso(150)

# Retirar $50
Alex.egreso(50)
```

# Reto de Alto Nivel — Sistema de Cajero Automático

## Objetivo

Desarrollar un sistema bancario utilizando programación
orientada a objetos y dos clases: Cliente y Cajero.

### Clase Cliente

Contendrá los atributos:
- nombre
- cantidad

Y los métodos:
- __init__()
- depositar()
- retirar()
- mostrar_todo()

### Clase Cajero

Administrará tres objetos de la clase Cliente.

Sus métodos serán:
- __init__()
- operar()
- deposito_total()

El programa permitirá realizar depósitos, retiros
y calcular el dinero depositado durante el día.

```python
# Crear la clase Cliente
class Cliente:

    def __init__(self, nombre):
        self.nombre = nombre
        self.cantidad = 0

    def depositar(self, monto):
        if monto <= 0:
            print("El depósito debe ser mayor a cero")
        else:
            self.cantidad += monto
            print(f"{self.nombre} depositó ${monto}")

    def retirar(self, monto):
        if monto <= 0:
            print("El retiro debe ser mayor a cero")
        elif monto > self.cantidad:
            print("No se tiene saldo suficiente")
        else:
            self.cantidad -= monto
            print(f"{self.nombre} retiró ${monto}")

    def mostrar_todo(self):
        print(f"Cliente: {self.nombre} | Saldo: ${self.cantidad}")
```

## Clase Cajero

### Objetivo

Crear una clase llamada Cajero que administre
tres objetos de la clase Cliente.

La clase tendrá los siguientes métodos:

- `__init__()`: inicializa el cajero con tres clientes.
- `operar()`: realiza depósitos y retiros.
- `deposito_total()`: calcula el dinero depositado
  durante el día.

El sistema permitirá consultar los saldos individuales
y el total de depósitos realizados.

```python
# Crear la clase Cajero
class Cajero:

    def __init__(self):
        self.cliente1 = Cliente("Alex")
        self.cliente2 = Cliente("Brenda")
        self.cliente3 = Cliente("Carlos")
        self.total_depositos = 0

    def operar(self):
        # Depósitos
        self.cliente1.depositar(100)
        self.cliente2.depositar(200)
        self.cliente3.depositar(300)

        self.total_depositos += 100 + 200 + 300

        # Retiros
        self.cliente1.retirar(50)
        self.cliente2.retirar(100)
        self.cliente3.retirar(400)

        # Mostrar saldos
        print("\n--- Saldos de clientes ---")
        self.cliente1.mostrar_todo()
        self.cliente2.mostrar_todo()
        self.cliente3.mostrar_todo()

    def deposito_total(self):
        print(f"\nTotal depositado durante el día: ${self.total_depositos}")
```

## Pruebas del sistema de cajero automático

### Objetivo

Comprobar el funcionamiento de las clases Cliente y Cajero.

Se realizarán las siguientes operaciones:

1. Crear un cajero con tres clientes.
2. Depositar $100 a Alex.
3. Depositar $200 a Brenda.
4. Depositar $300 a Carlos.
5. Realizar retiros de dinero.
6. Comprobar que no se permitan retiros sin saldo suficiente.
7. Mostrar los saldos finales de los clientes.
8. Calcular el total de depósitos realizados durante el día.

```python
# Crear un objeto de la clase Cajero
cajero = Cajero()

# Ejecutar depósitos y retiros
cajero.operar()

# Mostrar el total depositado durante el día
cajero.deposito_total()
```
