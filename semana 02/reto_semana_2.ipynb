## Semana 02 · Reto MIS2: Sucesión de Fibonacci en Python

**Objetivo:** desarrollar funciones que generen la sucesión de Fibonacci utilizando ciclos `for` y `while`, condiciones `if` y `else`, listas y operadores de comparación.

**Datos del ejercicio:** `fibonacci(9)` para el primer reto y `fibonacci(150)` para el segundo.

> Este notebook contiene los dos retos de la actividad MIS2. Ejecuta las celdas en orden y comprueba los resultados.


```python
def fibonacci(a):
    respuesta = 0
    lista = [0]

    for i in range(1, a + 1):
        if i == 1:
            respuesta = lista[-1] + 1
        else:
            respuesta = lista[-1] + lista[-2]

        lista.append(respuesta)
        print(respuesta)

    return respuesta
```

```python
fibonacci(9)
```

## Reto número 2 — Sucesión de Fibonacci con while

**Objetivo:** generar la sucesión de Fibonacci utilizando un ciclo `while` hasta obtener un número mayor que 150.

```python
def fibonacci(a):
    respuesta = 0
    lista = [0]

    while respuesta <= a:
        if respuesta == 0:
            respuesta = lista[-1] + 1
        else:
            respuesta = lista[-1] + lista[-2]

        lista.append(respuesta)
        print(respuesta)

    return respuesta
```

```python
fibonacci(150)
```
