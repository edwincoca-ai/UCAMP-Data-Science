**Objetivo:** utilizar variables y respetar la jerarquía de operaciones en Python.

**Datos del ejercicio:** `x = 60`, `y = 5.3`, `z = -3.8`.

> Este notebook contiene las expresiones de la actividad. Ejecuta las celdas en orden y comprueba los resultados. Las operaciones con `float` pueden mostrar pequeñas diferencias de precisión.

```python
# 1. Declaración de variables
x = 60       # int: número entero
y = 5.3      # float: número decimal
z = -3.8     # float: número decimal negativo

print(f"x = {x}, y = {y}, z = {z}")
```

### Incisos a)–d)

El símbolo `%` es el **módulo**: devuelve el residuo de una división, no un porcentaje.

```python
# 2. Primer bloque de operaciones
a = x + y + z
b = 2 * y + 3 * (x - z)
c = x / y
d = x % y

for inciso, resultado in [('a', a), ('b', b), ('c', c), ('d', d)]:
    print(f"{inciso}) {resultado}")
```

### Incisos e)–n)

**Observa:** `2 * x / 3 * y` se evalúa de izquierda a derecha, mientras que `2 * x / (3 * y)` calcula primero el denominador. Las expresiones no son equivalentes.

```python
# 3. Segundo bloque de operaciones: expresiones del reto
e = x / (y + z)
f = (x / y) + z
g = 2 * x / 3 * y
h = 2 * x / (3 * y)
i = x * y % z
j = x * (y % z)
k = 3 * x - z - 2 * x
l = 2 * x / 5 % y
m = x - 100 % y % z
n = x - y - z * 2

resultados = {'e': e, 'f': f, 'g': g, 'h': h, 'i': i, 'j': j,
              'k': k, 'l': l, 'm': m, 'n': n}
for inciso, resultado in resultados.items():
    print(f"{inciso}) {resultado}")
```

### Verificación y presentación

`round(valor, 2)` redondea la presentación numérica a dos decimales; no modifica el cálculo original. La representación binaria de los números `float` puede producir resultados como `201.99999999999997` en lugar de `202`.

```python
# 4. Resumen legible de todos los resultados
todos = {'a': a, 'b': b, 'c': c, 'd': d, **resultados}
print('RETO MIS1 — RESULTADOS (redondeados para lectura)')
print('-' * 45)
for inciso, valor in todos.items():
    print(f'{inciso}) {valor:.2f}')
```
