# Reto 1 — Análisis de datos con NumPy

## Objetivo

Importar un archivo de datos y utilizar NumPy para
realizar operaciones estadísticas.

Se calcularán:

1. Suma.
2. Valor mínimo.
3. Valor máximo.
4. Suma acumulada.
5. Media.
6. Mediana.
7. Desviación estándar.

```python
import numpy as np
```

```python
arr1 = np.loadtxt("reto1.txt")
print(arr1)
```

```python
print(type(arr1))
print(arr1.shape)
print(arr1.size)
```

## 3. Operaciones estadísticas con NumPy

### Objetivo

Aplicar funciones estadísticas a la primera columna
del arreglo para conocer el comportamiento de los datos.

Se calcularán:

1. Suma.
2. Valor mínimo.
3. Valor máximo.
4. Suma acumulada.
5. Media.
6. Mediana.
7. Desviación estándar.

Estas operaciones son fundamentales para el análisis
exploratorio de datos.

```python
# Suma
a = arr1[:, 0].sum()
print("Suma:", a)

# Mínimo
b = arr1[:, 0].min()
print("Mínimo:", b)

# Máximo
c = arr1[:, 0].max()
print("Máximo:", c)

# Suma acumulada
d = arr1[:, 0].cumsum()
print("Suma acumulada:", d)

# Media
e = arr1[:, 0].mean()
print("Media:", e)

# Mediana
f = np.median(arr1[:, 0])
print("Mediana:", f)

# Desviación estándar
g = arr1[:, 0].std()
print("Desviación estándar:", g)
```

# Reto 2 — Análisis de datos con Pandas

## Objetivo

Convertir el arreglo de NumPy en un DataFrame utilizando
un diccionario de Python.

El DataFrame tendrá tres columnas:

- A: primera columna del arreglo.
- B: segunda columna del arreglo.
- C: tercera columna del arreglo.

Posteriormente, se realizarán las mismas operaciones
estadísticas utilizadas con NumPy para comparar
los resultados de ambas bibliotecas.

## Herramientas

- Python
- NumPy
- Pandas

```python
# Importar la biblioteca Pandas
import pandas as pd

# Crear un diccionario con las columnas del arreglo
dic = {
    "A": arr1[:, 0],
    "B": arr1[:, 1],
    "C": arr1[:, 2]
}

# Convertir el diccionario en un DataFrame
df = pd.DataFrame(dic)

# Mostrar las primeras cinco filas
df.head()
```

## 2. Operaciones estadísticas con Pandas

### Objetivo

Aplicar operaciones estadísticas sobre la columna A
del DataFrame y comparar los resultados con NumPy.

Se calcularán las siguientes medidas:

1. Suma.
2. Valor mínimo.
3. Valor máximo.
4. Suma acumulada.
5. Media.
6. Mediana.
7. Desviación estándar.

### Comparación

NumPy permite realizar operaciones matemáticas sobre
arreglos numéricos, mientras que Pandas facilita el
análisis de datos organizados en filas y columnas.

```python
# Suma
a = df.A.sum()
print("Suma:", a)

# Mínimo
b = df.A.min()
print("Mínimo:", b)

# Máximo
c = df.A.max()
print("Máximo:", c)

# Suma acumulada
d = df.A.cumsum()
print("Suma acumulada:")
print(d)

# Media
e = df.A.mean()
print("Media:", e)

# Mediana
f = df.A.median()
print("Mediana:", f)

# Desviación estándar
g = df.A.std()
print("Desviación estándar:", g)
```

## Conclusión — Comparación entre NumPy y Pandas

Durante los dos primeros ejercicios del Reto MIS3 se
utilizaron las bibliotecas NumPy y Pandas para realizar
operaciones estadísticas sobre un conjunto de datos.

Con NumPy se trabajó directamente con un arreglo
bidimensional de 18 filas y 3 columnas, mientras que
con Pandas se creó un DataFrame utilizando un diccionario.

Ambas bibliotecas permitieron calcular la suma, el mínimo,
el máximo, la suma acumulada, la media y la mediana,
obteniendo los mismos resultados.

La desviación estándar presentó una diferencia debido
a que NumPy utiliza por defecto `ddof=0`, mientras que
Pandas utiliza `ddof=1`.

### Aprendizaje

NumPy facilita los cálculos numéricos con arreglos,
mientras que Pandas permite organizar, manipular y
analizar datos mediante tablas.

Estas herramientas son fundamentales para el análisis
exploratorio de datos y la ciencia de datos.
