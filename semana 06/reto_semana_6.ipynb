```python
import pandas as pd

archivo = "Reto de la semana 06.xlsx"

df = pd.read_excel(
    archivo,
    sheet_name="Reto Titanic",
    header=24,
    usecols="A:D"
)

df.head(10)
```

```python
# Revisar cantidad de registros y columnas
print("Dimensiones:", df.shape)

# Revisar tipos de datos
print("\nTipos de datos:")
print(df.dtypes)

# Identificar valores faltantes
print("\nValores faltantes:")
print(df.isnull().sum())
```

```python
# Contar los valores faltantes por columna
df.isnull().sum()
```

```python
# Estadísticas descriptivas de las edades

print("Promedio:", df["Age"].mean())
print("Mediana:", df["Age"].median())
print("Moda:", df["Age"].mode()[0])
print("Varianza:", df["Age"].var())
print("Desviación estándar:", df["Age"].std())
```

```python

# Seleccionar únicamente a los sobrevivientes
sobrevivientes = df[df["Survived"] == 1]

# Calcular sus estadísticas
print("Promedio:", sobrevivientes["Age"].mean())
print("Mediana:", sobrevivientes["Age"].median())
print("Moda:", sobrevivientes["Age"].mode()[0])
print("Varianza:", sobrevivientes["Age"].var())
print("Desviación estándar:", sobrevivientes["Age"].std())
```

```python

# Seleccionar pasajeros que no sobrevivieron
no_sobrevivientes = df[df["Survived"] == 0]

# Calcular sus estadísticas
print("Promedio:", no_sobrevivientes["Age"].mean())
print("Mediana:", no_sobrevivientes["Age"].median())
print("Moda:", no_sobrevivientes["Age"].mode()[0])
print("Varianza:", no_sobrevivientes["Age"].var())
print("Desviación estándar:", no_sobrevivientes["Age"].std())
```

```python

# Seleccionar únicamente a las mujeres
mujeres = df[df["Sex"] == 0]

# Calcular sus estadísticas
print("Promedio:", mujeres["Age"].mean())
print("Mediana:", mujeres["Age"].median())
print("Moda:", mujeres["Age"].mode()[0])
print("Varianza:", mujeres["Age"].var())
print("Desviación estándar:", mujeres["Age"].std())
```

```python

# Seleccionar únicamente a los hombres
hombres = df[df["Sex"] == 1]

# Calcular sus estadísticas
print("Promedio:", hombres["Age"].mean())
print("Mediana:", hombres["Age"].median())
print("Moda:", hombres["Age"].mode()[0])
print("Varianza:", hombres["Age"].var())
print("Desviación estándar:", hombres["Age"].std())
```

```python

# Seleccionar pasajeros de primera clase
primera_clase = df[df["Pclass"] == 1]

# Calcular sus estadísticas
print("Promedio:", primera_clase["Age"].mean())
print("Mediana:", primera_clase["Age"].median())
print("Moda:", primera_clase["Age"].mode()[0])
print("Varianza:", primera_clase["Age"].var())
print("Desviación estándar:", primera_clase["Age"].std())
```

```python

# Seleccionar pasajeros de tercera clase
tercera_clase = df[df["Pclass"] == 3]

# Calcular sus estadísticas
print("Promedio:", tercera_clase["Age"].mean())
print("Mediana:", tercera_clase["Age"].median())
print("Moda:", tercera_clase["Age"].mode()[0])
print("Varianza:", tercera_clase["Age"].var())
print("Desviación estándar:", tercera_clase["Age"].std())
```

```python

# Seleccionar mujeres sobrevivientes de primera clase
mujeres_primera = df[
    (df["Survived"] == 1) &
    (df["Sex"] == 0) &
    (df["Pclass"] == 1)
]

# Calcular sus estadísticas
print("Promedio:", mujeres_primera["Age"].mean())
print("Mediana:", mujeres_primera["Age"].median())
print("Moda:", mujeres_primera["Age"].mode()[0])
print("Varianza:", mujeres_primera["Age"].var())
print("Desviación estándar:", mujeres_primera["Age"].std())
```

```python

# Seleccionar hombres sobrevivientes de primera clase
hombres_primera = df[
    (df["Survived"] == 1) &
    (df["Sex"] == 1) &
    (df["Pclass"] == 1)
]

# Calcular sus estadísticas
print("Promedio:", hombres_primera["Age"].mean())
print("Mediana:", hombres_primera["Age"].median())
print("Moda:", hombres_primera["Age"].mode()[0])
print("Varianza:", hombres_primera["Age"].var())
print("Desviación estándar:", hombres_primera["Age"].std())
```

```python

# Seleccionar mujeres sobrevivientes de tercera clase
mujeres_tercera = df[
    (df["Survived"] == 1) &
    (df["Sex"] == 0) &
    (df["Pclass"] == 3)
]

# Calcular sus estadísticas
print("Promedio:", mujeres_tercera["Age"].mean())
print("Mediana:", mujeres_tercera["Age"].median())
print("Moda:", mujeres_tercera["Age"].mode()[0])
print("Varianza:", mujeres_tercera["Age"].var())
print("Desviación estándar:", mujeres_tercera["Age"].std())
```

```python

# Seleccionar hombres sobrevivientes de tercera clase
hombres_tercera = df[
    (df["Survived"] == 1) &
    (df["Sex"] == 1) &
    (df["Pclass"] == 3)
]

# Calcular sus estadísticas
print("Promedio:", hombres_tercera["Age"].mean())
print("Mediana:", hombres_tercera["Age"].median())
print("Moda:", hombres_tercera["Age"].mode()[0])
print("Varianza:", hombres_tercera["Age"].var())
print("Desviación estándar:", hombres_tercera["Age"].std())
```

```python

# Organizar los diez grupos analizados
grupos = {
    "General": df,
    "Sobrevivientes": sobrevivientes,
    "No sobrevivientes": no_sobrevivientes,
    "Mujeres": mujeres,
    "Hombres": hombres,
    "Primera clase": primera_clase,
    "Tercera clase": tercera_clase,
    "Mujeres sobrevivientes 1ra": mujeres_primera,
    "Hombres sobrevivientes 1ra": hombres_primera,
    "Mujeres sobrevivientes 3ra": mujeres_tercera,
    "Hombres sobrevivientes 3ra": hombres_tercera
}

# Calcular las medidas estadísticas de cada grupo
resultados = []

for nombre, grupo in grupos.items():
    edades = grupo["Age"]

    resultados.append({
        "Grupo": nombre,
        "Promedio": edades.mean(),
        "Mediana": edades.median(),
        "Moda": edades.mode().iloc[0],
        "Varianza": edades.var(),
        "Desviación estándar": edades.std()
    })

# Crear y mostrar la tabla
tabla_final = pd.DataFrame(resultados)
tabla_final.round(2)
```

```python

import matplotlib.pyplot as plt

# Crear gráfica de promedios
plt.figure(figsize=(12, 6))

plt.bar(
    tabla_final["Grupo"],
    tabla_final["Promedio"],
    color="steelblue"
)

plt.title("Edad promedio de los pasajeros del Titanic")
plt.xlabel("Grupos de pasajeros")
plt.ylabel("Edad promedio (años)")

plt.xticks(rotation=75, ha="right")
plt.tight_layout()
plt.show()
```


## Conclusiones del análisis estadístico

En este reto analizamos las edades de los pasajeros del Titanic mediante Python, Pandas y Matplotlib.

### Principales hallazgos

1. **Edad general:** El promedio de edad fue de 27.70 años y la mediana de 27.50 años.

2. **Supervivencia:** Los sobrevivientes tenían una edad promedio de 26.16 años, mientras que los no sobrevivientes registraron 29.34 años.

3. **Sexo:** Los hombres tenían una edad promedio de 28.10 años y las mujeres de 27.32 años.

4. **Clase del pasajero:** Los pasajeros de primera clase tenían una edad promedio de 36.22 años, frente a los 22.14 años de tercera clase.

5. **Sobrevivientes de primera clase:** Los hombres tenían una edad promedio de 43.60 años y las mujeres de 34.80 años.

6. **Sobrevivientes de tercera clase:** Las mujeres tenían una edad promedio de 24.80 años y los hombres de 22.00 años.

### Conclusión general

Los resultados muestran diferencias en las edades de los pasajeros según su sexo, clase y condición de supervivencia.

La clase del pasajero presentó diferencias importantes en las edades promedio. Sin embargo, estas estadísticas descriptivas no permiten establecer por sí solas relaciones de causa y efecto.
